## Concatenate Arrays

```python
c1=np.array([1,2,3])
c2=np.array([6,5,4])
np.concatenate([c1,c2])
```

```python
x=np.full((2,3),1)
y=np.full((2,3),9)
print(x)
print()
print(y)
```

```python
np.concatenate((x,y),axis=0)
```

```python
np.concatenate((x,y),axis=1)
```

```python
z=np.array([1,2,3])
z2=np.array([[1,2,3],[4,5,6]])
print(z)
print()
print(z2)
```

```python
np.vstack([z,z2])
```

```python
z=z.reshape(1,3)
z
```

## Split Arrays

```python
s=np.arange(10)
s
```

```python
s1,s2,s3=np.split(s,[3,6])
print(s1,s2,s3)
```

```python
grid=np.arange(16).reshape(4,4)
grid
```

```python
up,low=np.vsplit(grid,[2])
print(up)
print()
print(low)
```

```python
left,right=np.hsplit(grid,[2])
print(left)
print()
print(right)
```

## Ufuncs

```python
x=np.arange(4)
print("x = ",x)
print("add =",x+4)
print("div =",x/2)
print("floor div =",x//2)
print("modulo =",x%2)
print("power =",x**4)
print("negative =",-x)
print("negative =",np.negative(x))
print("absolute =",np.abs(x))
print("absolute =",np.absolute(x))
```
